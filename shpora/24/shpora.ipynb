```python
# файл содержит только десятичные цифры и буквы. Определите максимальное число,
# не превышающее 10**6, в котором сумма всех цифр, возведенная в степень,
# равная длине числа, равна самому числу. В ответе укажите количество
# повторений этого числа в файле.
def check(num: int):
    s = sum([int(i) for i in str(num)]) ** len(str(num))
    return s == num


text = open("146_1.txt").read()
mx = 0
for start in range(len(text) - 1):
    for end in range(start + 2, len(text)):
        sn = text[start:end]
        if sn.isdigit():
            n = int(sn)
            if check(n):
                mx = max(mx, n)
        else:
            break
print(mx)  # 2401
print(text.count("2401"))
```

```python
# файл состоит из T, U, V, W, X, Y и Z. определите максимальное количество
# символов в непрерывной подпоследовательности, состоящей из идущих подряд
# групп символов VWXYZ в указанном порядке. При этом в начале и в конце
# искомой последовательности группа символов VWXYZ может быть неполной.

from re import finditer

f = open("154_1.txt")
s = f.readline()
p = r"((VWXYZ|WXYZ|XYZ|YZ|Z){1}(VWXYZ)+(VWXYZ|VWXY|VWX|VW|V))"
p2 = rf"(?=({p}))"
res = []
for x in finditer(p2, s):
    res.append(len(x.group(1)))
print(max(res))
```

```python
# файл состоящий из всего алфавита латиницы ABC…..XYZ, а также десятичных цифр.
# Изначально будем считать что строка – это число записанное в 36-й системе счисления.
# Найдите максимальную длину подстроки в которой выполняются условия: если подстрока
# не имеет незначащих нулей, если подстроку перевести в 10-ю систему счисления –
# значение будет делиться на 36, а также если  в этой подстроке изначально было
# не больше 4 букв Z.

with open("1437_1.txt") as f:
    text = f.read().strip()
mx_len = 0
start = 0
z_count = 0
for end in range(len(text)):
    if text[end] == "Z":
        z_count += 1
    while z_count > 4:
        if text[start] == "Z":
            z_count -= 1
        start += 1
    if text[end] == "0":
        curr_start = start
        while curr_start < end and text[curr_start] == "0":
            curr_start += 1
        mx_len = max(mx_len, end - curr_start + 1)
print(mx_len)
```

```python
# файл содержит заглавные буквы пробелы и точки. Назовём предложением последовательность символов,
# которая не начинается с пробела, заканчивается точкой, перед которой стоит буква (не пробел),
# и не содержит внутри других точек. Словом будем называть последовательность латинских букв.
# Определите максимальную длину предложения, в котором слова расставлены в порядке невозрастания длины
# (второе слово не длиннее первого и т. д.).

import re

text = open("1753_1.txt").read()
m = re.findall(r"(?:[A-Z]+ +)*[A-Z]+\.", text)
res = []
for st in m:
    s = st[:-1].split()
    s = [len(i) for i in s]
    if s == sorted(s, reverse=True):
        res.append(st)
    else:
        for start in range(len(st)):
            if start == 0 or st[start] != " " and st[start - 1] == " ":
                s = st[start:-1].split()
                s = [len(i) for i in s]
                if s == sorted(s, reverse=True):
                    res.append(st[start:])
print(len(max(res, key=len)))
```

```python
# файл содержит заглавные буквы и десятичные цифры. Найдите минимальное количество идущих
# подряд символов, среди которых есть все цифры и есть ровно три буквы латинского алфавита
# от A до F (других букв может быть сколько угодно)


from string import printable

text = open("1862_1.txt").readline()
letters = dict.fromkeys(printable, 0)
start = 0
mn = 10**6
for end in range(len(text)):
    letters[text[end]] += 1
    while (
        all([letters[i] > 0 for i in "0123456789"])
        and sum([letters[i] for i in "ABCDEF"]) >= 3
    ):
        if (
            all([letters[i] > 0 for i in "0123456789"])
            and sum([letters[i] for i in "ABCDEF"]) == 3
        ):
            mn = min(mn, end - start + 1)
        letters[text[start]] -= 1
        start += 1

print(mn)
```

```python
# файл состоит из цифр 0–9 и букв A–Z. Определите в файле длину самой длинной непрерывной
# подпоследовательности, которая состоит только из цифр и содержит не более 3 различных цифр

with open("1948_1.txt") as f:
    s = f.read().strip()

max_len = 0
start = 0
digit_counts = {}

for end in range(len(s)):
    char = s[end]
    if not char.isdigit():
        digit_counts.clear()
        start = end + 1
        continue
    digit_counts[char] = digit_counts.get(char, 0) + 1
    while len(digit_counts) > 3:
        left_char = s[start]
        digit_counts[left_char] -= 1
        if digit_counts[left_char] == 0:
            del digit_counts[left_char]
        start += 1
    max_len = max(max_len, end - start + 1)
print(max_len)
```
